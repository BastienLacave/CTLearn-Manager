```python
import astropy.units as u

from ctlearn_manager import CTLearnManagerProject, DataSample
from ctlearn_manager.utils import ClusterConfiguration, ParticleType, Cuts, DefaultCuts
```

# Note on `Cuts`

The Manager allows you to apply cuts to the data through the `Cuts`class, or the `DefaultCuts` which contains premade sets of cuts, such as global 0.9 gammaness cut or 70% efficiency cut.

Note that in order to use efficiency based of sensitivity imptimized cuts (therefore energy-dependent cuts), you need to have produced correspondin IRFs and cuts files, with a config file that corresponds to the cuts you require.

```python
project_directory = "/path/to/your/project"
manager_project = CTLearnManagerProject(project_directory)
tri_model = manager_project.open_tri_model(f"tri_model_nickname")
tri_model.get_available_MC_directions()
```

```python
zenith, azimuth = 10 * u.deg, 180 * u.deg
```

```python
tri_model.plot_ROC_curve_DL2(zenith, azimuth, 5)
```

```python
tri_model.plot_angular_resolution_DL2(
    [zenith.value] * u.deg, [azimuth.value] * u.deg, cuts=[DefaultCuts.EFF_70.value]
)
tri_model.plot_energy_resolution_DL2(
    [zenith.value] * u.deg, [azimuth.value] * u.deg, cuts=[DefaultCuts.EFF_70.value, Cuts(gammaness_cut=0.9)]
)
```

```python
tri_model.plot_migration_matrix(
    zenith, azimuth, particle_types=[ParticleType.GAMMA_POINT, ParticleType.PROTON]
)
tri_model.plot_DL2_AltAz(
    zenith, azimuth, particle_types=[ParticleType.GAMMA_POINT, ParticleType.PROTON]
)
tri_model.plot_DL2_classification(
    zenith,
    azimuth,
    particle_types=[
        ParticleType.GAMMA_POINT,
        ParticleType.PROTON,
        ParticleType.ELECTRON,
    ],
)
tri_model.plot_DL2_energy(zenith, azimuth)
```
