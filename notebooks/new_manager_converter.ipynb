```python
import os
import glob
```

```python
old_models_directory = "/capstor/store/cscs/cta/cta03/ctlearn_blacave/CTLearn_Models/"
new_project_directory = "/users/blacave/PhD/New_Manager_Test/SingleModelProject"
```

```python
model_list = glob.glob(f"{old_models_directory}/*")
print(model_list)
recos = ['cameradirection', 'energy', 'type']

for reco in recos:
    old_model_name = f"LST1_cleanedTRN_allnodes_{reco}"
    new_tri_model_name = f"SingleModel" # Should NOT include the reco task (will be added automatically)

    
    input_dir = f"{old_models_directory}/{old_model_name}"
    if reco in ['cameradirection', 'skydirection']:
        reco = 'direction'

    # DO NOT CHANGE THE FOLLOWING LINES #####################################################################
    output_dir = f"{new_project_directory}/models/{new_tri_model_name}/{reco}/{new_tri_model_name}_{reco}/" #
    os.makedirs(output_dir, exist_ok=True)                                                                  #
    cmd = f"scp -r {input_dir}/* {output_dir}"   
                                                                                                                
    # !{cmd} # UNCOMMENT THIS LINE TO ACTUALLY COPY THE FILES 
                                                                                                        
    directories = glob.glob(output_dir + '*/')                                                              
    for direct in directories:                                                                              
        cmd = f"mv {direct} {direct.replace(old_model_name, f'{new_tri_model_name}_{reco}')}" 

        # !{cmd} # UNCOMMENT THIS LINE TO ACTUALLY COPY THE FILES        
```
