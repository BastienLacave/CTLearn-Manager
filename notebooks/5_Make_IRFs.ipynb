```python
import astropy.units as u

from ctlearn_manager.io.io import load_model_from_index
from ctlearn_manager.tri_model import CTLearnTriModelManager
from ctlearn_manager.utils import ParticleType
```

```python
MODEL_INDEX_FILE = "/path/to/your/ctlearn_models_index.h5"
energy_model = load_model_from_index("energy_model_nickname", MODEL_INDEX_FILE)
direction_model = load_model_from_index("direction_model", MODEL_INDEX_FILE)
type_model = load_model_from_index("type_model", MODEL_INDEX_FILE)
Tri_Model = CTLearnTriModelManager(
    direction_model=direction_model, energy_model=energy_model, type_model=type_model
)
```

# 🔀 If you have multiple testing files for same position, merge them

```python
Tri_Model.merge_DL2_files(
    10.0 * u.deg,
    102.199 * u.deg,
    output_file="/path/to/your/merged.h5",
    particle_type=ParticleType.GAMMA_POINT,
    overwrite=True,
)
```

# 📡 Produce the IRFs

```python
zenith, azimuth = 10 * u.deg, 180 * u.deg

# ⚠️⚠️⚠️ Make sure output files are unique !!!
details = "_eff70" # To make sure every file has a unique name based on the cuts implemented.
config = f"/path/to/your/public-conf{details}.yml"  # Example config found in ressources, do not modify or move it after producing IRFs
output_cuts_file = (
    f"/path/to/your/IRFs/cuts_{zenith.value:.2f}_{azimuth.value:.2f}{details}.fits"
)
output_irf_file = f"/path/to/your/IRFs/IRFs_{zenith.value:.2f}_{azimuth.value:.2f}{details}.fits"
output_benchmark_file = (
    f"/path/to/your/IRFs/benchmark_{zenith.value:.2f}_{azimuth.value:.2f}{details}.fits"
)
```

```python
Tri_Model.produce_irfs(
    zenith,
    azimuth,
    config=config,
    output_cuts_file=output_cuts_file,
    output_irf_file=output_irf_file,
    output_benchmark_file=output_benchmark_file,
    pointlike=True,
    electrons=False,
    protons=True,
)
```

# 📉 Plot benchmark and IRFs

```python
Tri_Model.plot_benchmark(zenith, azimuth)
```
