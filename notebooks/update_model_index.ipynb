```python
import astropy.units as u

from ctlearn_manager.io.io import load_model_from_index
from ctlearn_manager.tri_model import CTLearnTriModelManager
from ctlearn_manager.utils import ParticleType, remove_table_from_h5
```

1. Delete the MC/DL2 table
2. Launch testing with overwrite set to False, so that only the MC/DL2 table is recreated with the desired files
3. Merge these files, even if there is one per pointing direction.

# 1. Delete tables

```python
MODEL_INDEX_FILE = "path"
model_nickname = "model_nickname"
particle_type = ParticleType.GAMMA_POINT
```

```python
remove_table_from_h5(MODEL_INDEX_FILE, f"/{model_nickname}/DL2/MC/{particle_type.value}")
remove_table_from_h5(MODEL_INDEX_FILE, f"/{model_nickname}/DL2/MC/{particle_type.value}.__table_column_meta__") # removing metadata is mandatory
```

# 2. Recreate DL2 Tables

```python
MODEL_INDEX_FILE = "/path/to/your/ctlearn_models_index.h5"
energy_model = load_model_from_index("energy_model_nickname", MODEL_INDEX_FILE)
direction_model = load_model_from_index("direction_model", MODEL_INDEX_FILE)
type_model = load_model_from_index("type_model", MODEL_INDEX_FILE)
Tri_Model = CTLearnTriModelManager(
    direction_model=direction_model, energy_model=energy_model, type_model=type_model
)
Tri_Model.get_available_testing_directions()
```

```python
# Copy here the direction you want to update in the manager
# ⚠️ Make sure the output files already exist (the ones tat were in the tables you deleted above).
zenith, azimuth = 32.059 * u.deg, 248.099 * u.deg
```

```python
Tri_Model.launch_testing(
    zenith,
    azimuth,
    output_dirs=["/path/to/your/Testing_models/gamma_point/"],
    launch_particle_types=[
        ParticleType.GAMMA_POINT
    ],  # Add as much as you want, must provide a directory for each
    config_dir="/path/to/your/configs/",  # Where the .sh scripts, configs and logs will be stored
    batch_size=64,
    overwrite=False
    # Other options are available
)
```

# 3. Merge (mandatory)

```python
Tri_Model.merge_DL2_files(
    zenith,
    azimuth,
    output_file="/path/to/your/merged/gamma_point_50_300E3GeV_20_20deg.h5", # If there is only one file, this output file is not written.
    particle_type=ParticleType.GAMMA_POINT,
    overwrite=True,
)
```
